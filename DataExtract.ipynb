```python
# def MoveFolder(start_dest, end_dest):
#   source_folder = "path/to/source_directory"
#   destination_folder = "path/to/destination_directory"

#   # Merge the folders seamlessly
#   shutil.copytree(source_folder, destination_folder, dirs_exist_ok=True)


def merge_folders_safely(src, dst):
    # Ensure the destination folder exists
    os.makedirs(dst, exist_ok=True)
    
    for root, dirs, files in os.walk(src):
        # Determine the relative path to maintain subfolder structure
        rel_path = os.path.relpath(root, src)
        dest_dir = os.path.join(dst, rel_path)
        os.makedirs(dest_dir, exist_ok=True)
        
        for file in files:
            src_file = os.path.join(root, file)
            dest_file = os.path.join(dest_dir, file)
            
            # If file exists, rename it sequentially
            if os.path.exists(dest_file):
                name, ext = os.path.splitext(file)
                counter = 1
                while os.path.exists(dest_file):
                    new_name = f"{name}_{counter}{ext}"
                    dest_file = os.path.join(dest_dir, new_name)
                    counter += 1
            
            shutil.copy2(src_file, dest_file)

# Usage
# merge_folders_safely("path/to/source", "path/to/destination")
```
